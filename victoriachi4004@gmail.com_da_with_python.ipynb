# INTRODUCTION TO MATPLOTLIB IN PYTHON.
WHAT IS MATPLOTLIB? 
## Matplotlib is a Python library that is used for Data Visualization, for creating Graphs, charts, and plots from data. It is also a field in data science that works alongside Numpy and Pandas.

    FEATURES OF A MATPLOTLIB
1. Easy to understand and use as a beginner
2. Highly customizable(colors, labels, styles, etc.)
3. Works well with numerical data
4. It can be used to create publication-quality graphs.
# Some functions () that you will come across in matplotlib include:
1. Groupby (): Groups similar data together.
2. Sum (): Add the values.
3. plot (): Creates the chart.
4. plt.title (): Add the chart title.
5. plt.xlabel (): Names the horizontal axis.
6. plt.ylabel (): Names the vertical axis.
7. plt.show (): Display the charts. etc

    ## TYPE OF CHARTS YOU CAN CREATE USING THE MATPLOTLIB
1. Bar chart.
2. Line chart.
3. Histogram.
4. Scatter.
5. Pie chart. etc
    

```python
import matplotlib as plt
```

```python
import matplotlib.pyplot as plt

# 1. Bar Chart: Revenue by Brand
brand_revenue = df.groupby('Brand')['Revenue_USD'].sum()
plt.figure(figsize=(6, 4))
plt.bar(brand_revenue.index, brand_revenue.values, color='Blue')
plt.title('Total Revenue by Brand')
plt.xlabel('Brand')
plt.ylabel('Revenue (USD)')
plt.xticks(rotation=30)
plt.show()

```

```python
## --- Histogram 

plt.hist(df['Price_USD'], bins=10, color='skyblue', edgecolor='black')
plt.title('Distribution of Shoe Prices')
plt.show()
```

```python
# --- Pie Chart 

counts = df['Sales_Channel'].value_counts()
plt.pie(counts, labels=counts.index, autopct='%1.1f%%')
plt.title('Sales by Channel')
plt.show()
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\USER\OneDrive\Documents\anaconda_projects\db\shoes_sales_dataset.csv")


```

```python
## The Count plot Is Used to categorize the data.
## Which shoe brand appears the most for each category
sns.countplot(x='Brand', data=df)
plt.xticks(rotation=45)
plt.show()
```

```python
## Creating the Histogram Plot using the seaborn function

sns.histplot(df['Price_USD'], bins=20, kde=True)
plt.show()
```
